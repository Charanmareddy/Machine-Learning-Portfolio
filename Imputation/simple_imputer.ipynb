```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
```

```python
x = pd.DataFrame({
    'col_1': [1.2, 2.5, None, 4.1, None, 6.3, None, 8.4, 9.1, 10.2],
    'col_2': [10, None, 30, 40, 50, None, 70, 80, 90, 100],
    'col_3': ['A', 'B', 'A', None, 'C', 'A', 'C', 'A', None, 'B'],
    'col_4': ['AA', 'BB', 'AA', np.nan, 'CC', 'AA', np.nan, 'AA', np.nan, 'BB']

})
```

```python
x
```

```python
x.info()
```

```python
imputer =SimpleImputer(strategy='constant',fill_value='unknow') # it will work for both 
imputer.fit_transform(x)
```

```python
imputer =SimpleImputer(strategy='constant',fill_value='unknow') 
imputer.fit_transform(x[['col_3','col_4']])
```

```python
imputer =SimpleImputer(strategy='mean') 
imputer.fit_transform(x[['col_1','col_2']])
```

```python
imputer =SimpleImputer(strategy='median') 
imputer.fit_transform(x[['col_1','col_2']])
```
