```python
import pandas as pd
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.pipeline import Pipeline
```

```python
df = pd.read_csv("IMDB Dataset.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
x = df.review
y = df.sentiment
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,train_size=0.8,random_state = 42)
```

```python
vectorizer = CountVectorizer()
```

```python
values = vectorizer.fit_transform(xtrain)
# values.toarray()
```

```python
names = vectorizer.get_feature_names_out()
names
```

```python
logistic_regression = LogisticRegression()
```

```python
pipeline = Pipeline(
            steps= [
                ('vectorization',vectorizer),
                ('model',logistic_regression)
            ]
            )
```

```python
model = GridSearchCV(
                estimator= pipeline,
                param_grid={
                    'vectorization__max_features':[10000,20000,None],
                    'vectorization__stop_words':[None,'english'],
                    'model__penalty':['l1','l2'],
                    'model__solver':['lbfgs','saga'],
                    'model__C':[0.1,1]
                },
                n_jobs=-1,
                verbose = 10,
                cv =3
)
```

```python
model.fit(xtrain,ytrain)
```

```python
model.cv_results_
```

```python
model.best_estimator_
```

```python
model.best_params_
```

```python
results = pd.DataFrame(model.cv_results_)
```

```python
results.sort_values(by='rank_test_score')
```

```python
y_pred_train = model.predict(xtrain)
```

```python
y_pred_test = model.predict(xtest)
```

```python
print(classification_report(y_pred_train,ytrain))
```

```python
print(classification_report(y_pred_test,ytest))
```
